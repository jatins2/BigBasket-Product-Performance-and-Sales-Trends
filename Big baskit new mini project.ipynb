# Step 8: Find out the outliers from the dataset according to the columns and fill them with the mean

```python
DF1=DF.select_dtypes([float])
DF1
```

```python
def outliers(DF1, col):
    q2=DF1[col].mean()
    q1=DF1[col].quantile(0.25)
    q3=DF1[col].quantile(0.75)
    iqr=q3-q1
    lower_limit=max(q1-1.5*iqr,0)
    upper_limit=q3+1.5*iqr
    return lower_limit , upper_limit

a=['sale_price','market_price','rating','discount']
for i in a:
    print (outliers(DF1,i))

    
```
